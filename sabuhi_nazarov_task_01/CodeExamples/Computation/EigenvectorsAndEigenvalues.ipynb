# Eigenvectors And Eigenvalues

## Linear Transformation

```python
import numpy as np

A = np.array([[2, 1], 
              [1, 3]])
v = np.array([1, 2])
transformed_vect = A @ v

print("Original vector:", v)
print("\nTransformed vector:", transformed_vect)
```

## Eigenvectors and Eigenvalues Computation

### `np.linalg.eig(A)` computes the eigenvalues and eigenvectors of matrix A.

```python
A = np.array([[4, -2], 
              [1,  1]])
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

## Eigendecomposition

### Reconstructing A Using Q * Λ * Q⁻¹ 

```python
A = np.array([[4, -2], 
              [1,  1]])

eigenvalues, eigenvectors = np.linalg.eig(A)
Q = eigenvectors  # Eigenvectors matrix
Λ = np.diag(eigenvalues)  # Diagonal matrix of eigenvalues
A_reconstructed = Q @ Λ @ np.linalg.inv(Q)  # Reconstruct A

print("Original Matrix:\n", A)
print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
print("\nReconstructed Matrix:\n", A_reconstructed)
```
