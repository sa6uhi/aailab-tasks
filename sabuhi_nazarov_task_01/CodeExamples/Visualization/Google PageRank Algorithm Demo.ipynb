# Google PageRank Algorithm

### Firstly, let's create simple network with 4 nodes

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()
edges = [(0, 1), (1, 2), (2, 0), (2, 3), (3, 2)]
G.add_edges_from(edges)
pagerank_scores = nx.pagerank(G)

plt.figure(figsize=(6, 4))
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color="lightblue",
        node_size=[v * 3000 for v in pagerank_scores.values()])

print("PageRank Scores:\n\n", pagerank_scores)

plt.show()
```

### Now, let's increase number of nodes

```python
G = nx.DiGraph()
edges = [
    (0, 1), (0, 2), (1, 2), (1, 3), (2, 0), (2, 3), (2, 4), 
    (3, 4), (4, 1), (4, 5), (5, 2), (5, 6), (6, 4), (6, 0)
]
G.add_edges_from(edges)
pagerank_scores = nx.pagerank(G)

plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="lightblue", 
        node_size=[v * 4000 for v in pagerank_scores.values()], 
        edge_color="gray", font_size=10)

print("PageRank Scores:", pagerank_scores)

plt.show()
```
